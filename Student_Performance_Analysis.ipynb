```python

import zipfile
import os

# Extract the inner ZIP file
with zipfile.ZipFile("student_performance.zip", "r") as outer:
    outer.extract("student.zip", ".")

# Extract the actual dataset
with zipfile.ZipFile("student.zip", "r") as inner:
    inner.extractall("student_data")

# Find the mathematics dataset
for root, dirs, files in os.walk("student_data"):
    for file in files:
        print(os.path.join(root, file))
```

```python

import pandas as pd

# Load dataset
df = pd.read_csv("student_data/student-mat.csv", sep=";")

# Display dataset information
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Show first five rows
display(df.head())
```

: Basic Data Analysis

```python

# Dataset overview
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nStatistical Summary:")
display(df.describe())

print("\nFinal Grade Distribution:")
print(df["G3"].value_counts().sort_index())
```

```python

# Dataset overview
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nStatistical Summary:")
display(df.describe())

print("\nFinal Grade Distribution:")
print(df["G3"].value_counts().sort_index())
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Final Grade Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["G3"], bins=15, kde=True)
plt.title("Distribution of Final Student Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()

# 2. Study Time vs Final Grade
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Weekly Study Time")
plt.ylabel("Final Grade")
plt.show()

# 3. Absences vs Final Grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.show()

# 4. Gender vs Final Grade
plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Gender vs Final Grade")
plt.xlabel("Gender")
plt.ylabel("Final Grade")
plt.show()
```

```python

print("Average Final Grade:", round(df["G3"].mean(), 2))

print("\nAverage Grade by Study Time:")
print(df.groupby("studytime")["G3"].mean())

print("\nAverage Grade by Gender:")
print(df.groupby("sex")["G3"].mean())

print("\nCorrelation between Absences and Final Grade:")
print(round(df["absences"].corr(df["G3"]), 3))
```

```python

# ============================================================
# STUDENT PERFORMANCE ANALYSIS
# Complete Exploratory Data Analysis Project
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")

# 1. LOAD DATASET
df = pd.read_csv("student_data/student-mat.csv", sep=";")

os.makedirs("project_outputs", exist_ok=True)

print("=" * 50)
print("STUDENT PERFORMANCE ANALYSIS")
print("=" * 50)

# 2. DATASET OVERVIEW
print("\n1. DATASET OVERVIEW")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

# 3. STATISTICAL ANALYSIS
print("\n2. STATISTICAL SUMMARY")
display(df.describe())

print("\nAverage Final Grade:", round(df["G3"].mean(), 2))
print("Highest Final Grade:", df["G3"].max())
print("Lowest Final Grade:", df["G3"].min())

# 4. GRADE DISTRIBUTION
plt.figure(figsize=(9, 5))
sns.histplot(df["G3"], bins=15, kde=True)
plt.title("Distribution of Final Student Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.savefig("project_outputs/grade_distribution.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 5. STUDY TIME VS PERFORMANCE
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Weekly Study Time Category")
plt.ylabel("Final Grade")
plt.savefig("project_outputs/studytime_vs_grade.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 6. ABSENCES VS PERFORMANCE
plt.figure(figsize=(9, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade")
plt.savefig("project_outputs/absences_vs_grade.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 7. GENDER VS PERFORMANCE
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Gender vs Final Grade")
plt.xlabel("Gender (F = Female, M = Male)")
plt.ylabel("Final Grade")
plt.savefig("project_outputs/gender_vs_grade.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 8. PARENTAL EDUCATION VS PERFORMANCE
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="Medu", y="G3")
plt.title("Mother's Education vs Final Grade")
plt.xlabel("Mother's Education Level")
plt.ylabel("Final Grade")
plt.savefig("project_outputs/mothers_education_vs_grade.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 9. CORRELATION HEATMAP
plt.figure(figsize=(12, 8))
correlation = df.select_dtypes(include="number").corr()
sns.heatmap(correlation, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.savefig("project_outputs/correlation_heatmap.png",
            dpi=300, bbox_inches="tight")
plt.show()

# 10. GROUP ANALYSIS
print("\n3. AVERAGE GRADE BY STUDY TIME")
display(df.groupby("studytime")["G3"].mean().round(2))

print("\n4. AVERAGE GRADE BY GENDER")
display(df.groupby("sex")["G3"].mean().round(2))

print("\n5. AVERAGE GRADE BY SCHOOL")
display(df.groupby("school")["G3"].mean().round(2))

print("\n6. ABSENCES AND FINAL GRADE CORRELATION")
print(round(df["absences"].corr(df["G3"]), 3))

# 11. SAVE ANALYSIS RESULTS
df.describe().to_csv("project_outputs/statistical_summary.csv")
correlation.to_csv("project_outputs/correlation_matrix.csv")

print("\n" + "=" * 50)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("Graphs and reports saved in project_outputs/")
print("=" * 50)
```
